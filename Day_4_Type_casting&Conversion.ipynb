## **Type Conversion**
**Type conversion** in Python is the process of changing a value from one data type to another. It ensures that operations are executed correctly and is split into two categories: 
1. Implicit Conversion (automatic) 
2. Explicit Conversion (manual casting).

## **1. Implicit Type Conversion (Automatic)**
Python automatically converts one data type to another at runtime when mixing different numeric types. 

```python
# Python promotes the int to a float automatically
integer_num = 10     # int
float_num = 5.5      # float

result = integer_num + float_num
print(result)        # Output: 15.5
print(type(result))  # Output: <class 'float'>
```

## **2. Explicit Type Conversion (Type Casting)**
Explicit conversion requires you to manually force a value to a specific type using built-in functions. This is critical when working with user interfaces or functions like input(), which ingest data strictly as strings.

```python
age = input("Enter your Age: ")
print(age)
print(type(age))
```

```python
# corrct and safe way for this to handle 
age = int(input("Enter your Age: "))
print(age)
print(type(age))
```

```python
no1 = input("Enter 1st Number: ")
no2 = input("Enter 2nd Number: ")

add = no1 + no2
print(add)
print(type(add))
```
