## **Loop Control Statements**
You can alter the flow of a loop using three keywords:

**1.**  break: 
* Exits the loop
* USeful when a certain condition is met and further processing in unnecessary.

**2.**  continue: 
* Skips the current iteration and moves to the next one.
* Ideal for bypassing specific conditions without stopping the entire loop.

**3.**  else: 
* Runs when the loop completes naturally end
* Can be used to execute code after the loop finishes processing.


```python
for i in range(10):
    if i == 3:
        print("Skipping 3")
        continue  # Skip the rest of the loop when i is 3
    if i == 8:
        print("Breaking the loop at number 8")
        break  # Exit the loop when i is 8
    print(f"Current number: {i}")
else:
    print("Loop completed without break.")
```

```python
for i in range(1, 1001, 100):
    
    if i == 501:
        continue
    print(i)
```

```python
for num in range(5):
    if num == 4:
        break     # Stops the loop at 4
    print(num)
```

```python
for num in range(5):
    if num == 2:
        continue  # Skips printing 2
    print(num)
```

## **Loop with an else Block**
Python allows an optional else block after a loop. The else block executes only if the loop finishes naturally without encountering a break statement.

```python
for x in range(3):
    print(x)
else:
    print("Loop finished successfully!")
```

## **Accessing Indexes with enumerate()**
If you need both the item and its index position while looping over a collection, wrap the collection in enumerate().

```python
items = ["A", "B", "C"]
for index, value in enumerate(items):
    print(f"Index {index} has value {value}")
```
