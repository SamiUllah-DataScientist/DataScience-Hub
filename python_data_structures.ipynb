## **Data Structures in Python**

**Python** provides **four** built-in data structures like **lists**, **dictionaries**, **sets**, and **tuples**—alongside powerful advanced structures in its standard library and the ability to build custom user-defined structures.

```python
## Python Collections as a Data Structure
# Used to store multiple values in one variable

# List: Ordered and changeable. Allows duplicate items.
# Tuple: Ordered and unchangeable. Allows duplicate items.
# Set: Unordered and unindexed. Contains unique values only (no duplicates).
# Dictionary: Stores data in key-value pairs.
```

## **1. List**
**List** is an Ordered, mutable collections that allow duplicate elements. It is enclosed in square brackets [ ].


```python
empty_list = []
print(empty_list)  # Output: []
```

```python
fruits = ["Apple", "Grapes", "Banana", "Orange"]

print(fruits)
print(type(fruits))
```

```python
numbers = [1, 2, 3, 4, 5]
mixed_list = [1, "Hello", 3.14, True]

print("numbers:", numbers)
print("mixed_list:", mixed_list)
```

 * Adding, Removing and Modifying Elements

```python
print("Original numbers:", numbers)

# Adding elements to the list
numbers.append(6)  # Adds 6 at the end of the list
print("Numbers after appending 6:", numbers)
numbers.insert(2, 2.5)  # Inserts 2.5 at index 2
print("Numbers after inserting 2.5 at index 2:", numbers)
```

```python
# Removing elements from the list
numbers.remove(6)  # Removes the first occurrence of 3
print("Numbers after removing 3:", numbers)
```

```python
removed_element = numbers.pop()  # Removes the last element and returns it
print("Removed element:", removed_element)
```

```python
# modifying list element
numbers[0] = 10  # Changes the first element to 10
print("Numbers after modifying the first element:", numbers)
```

## **2. Tuple**
**Tuple** is an Ordered, immutable collections that allow duplicate elements.

```python
empty_tuple = ()
print(empty_tuple)  # Output: ()

single_element_tuple = (1,)  # Note the comma
print(single_element_tuple)  # Output: (1,)

multiple_elements_tuple = (1, 2, 3, "Hello", 3.14)
print(multiple_elements_tuple)
```

```python
colors = ("Red","Green","Blue")

print(colors)
print(type(colors))
```

### **Immutability**

```python
# Trying to modify a tuple will rasise an error
try:
    colors[0] = "Yellow"  # This will raise a TypeError
except TypeError as e:
    print("Error:", e)
```

### **Unpacking a Tuple**

```python
person_info = ("Sami", 25, "New York")
name, age, city = person_info  # Unpacking the tuple
print("Name:", name)
print("Age:", age) 
print("City:", city)
```

## **3 Set**
**Set** is an Unordered collections of unique elements with no duplicates allowed.

```python
empty_set = set()
print(empty_set)  # Output: set()
```

```python
numbers = {1, 3, 5, 6, 6, 5, 3, 8, 9, 0}

# print only repeat numbers first time
print(numbers)
print(type(numbers))
```

* Adding and Removing Elements

```python
numbers.add(10)  # Adds 10 to the set
print("Numbers after adding 10:", numbers)
# numbers.remove(3)  # Removes 3 from the set
print("Numbers after removing 3:", numbers)
numbers.discard(100)  # Does nothing if 100 is not in the set
print("Numbers after discarding 100:", numbers)
```

```python
set_A = {1, 2, 3, 4}
set_B = {3, 4, 5, 6}

# union of two sets
print("Union of A and B:", set_A.union(set_B))  # Output: {1, 2, 3, 4, 5, 6}
# intersection of two sets 
print("Intersection of A and B:", set_A.intersection(set_B))  # Output: {3, 4}
# Difference of two sets
print("Difference of A and B:", set_A.difference(set_B))  # Output: {1, 2}
# symmetric difference of two sets: elements in either set_A or set_B but not in both
print("Symmetric Difference of A and B:", set_A.symmetric_difference(set_B))  #
```

## **4. Dictionary**
**Dictionary** is an Unordered (but preserves insertion order in Python 3.7+), mutable mappings of unique keys to values.

```python
empty_dict = {}
print("Empty dictionary:", empty_dict)
```

```python
intro_dict = {
    'name' : "Sami",
    'skill' : "Python",
    'age' : 18
}

print(intro_dict)
print(type(intro_dict))
```

```python
dict_cont = dict(name="Sami", skill="Python", age=18)
print(dict_cont)
```

```python
# Using curly braces
user = {
    "name": "Alice",
    "age": 28,
    "city": "New York"
}

print(user)
# Using the dict() constructor
user = dict(name="Alice", age=28, city="New York")

# Empty dictionary
empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
# Accessing values by keys

print("Name:", user["name"])  # Output: Alice
print("Age:", user.get("age"))    # Output: 28
```

```python
print("Keys:", user.keys())  # Output: dict_keys(['name', 'age', 'city'])
print("Values:", user.values())  # Output: dict_values(['Alice', 28, 'New York'])
print("Items:", user.items())  # Output: dict_items([('name', 'Alice'), ('age', 28), ('city', 'New York')])
```
