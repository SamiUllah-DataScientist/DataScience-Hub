## **Conditional Statements**
Python allow you to execute different blocks of code based on whether a specific condition evaluates to True or False. They are the core tool for controlling the flow and decision-making logic of your programs.

### **1. The Standard if Statement**

The most basic control structure. The code block executes only if the condition is true.


````python
age = 20
if age >= 18:
    print("You are eligible to vote.")  # Executes because condition is True
````

### **2. The if-else Statement**
Provides an alternative block of code that runs if the initial if condition evaluates to False.

````python
score = 73
if score >= 50:
    print("You passed!")
else:
    print("You failed.")  # Executes because score is less than 50
````

````python
rain = bool(input("Status(True/False): "))
if rain == True:
    print("It is Raining Now.")
else:
    print("NO it is not a raining.")
````

### **3. The if-elif-else Chain**
Used to check multiple conditions sequentially. Python evaluates them from top to bottom; as soon as one condition is true, its block runs, and the rest of the chain is skipped.

````python
traffic_light = "yellow"

if traffic_light == "red":
    print("Stop")
elif traffic_light == "yellow":
    print("Slow down")  # Executes
else:
    print("Go")
````

````python
marks = 56

if marks >= 90 and marks <= 100:
    print("Your Grade is A+")
elif marks >= 80 and marks < 90:
    print("Your Grade is A")
elif marks >= 70 and marks < 80:
    print("Your Grade is B")
elif marks >= 60 and marks < 70:
    print("Your Grade is C")
else:
    print("You are Failed. Try Again...")
````

### **4. Nested Conditions**
You can place conditional statements inside other conditional statements to check for sub-conditions.

````python
has_ticket = True
is_vip = False

if has_ticket:
    if is_vip:
        print("Welcome to the VIP lounge.")
    else:
        print("Welcome to the standard seating.")  # Executes
else:
    print("Please purchase a ticket.")
````

### **5. Inline Conditionals (Ternary Operator)**
A shorthand syntax to assign values or return results in a single line.

Syntax: python
```
value_if_true if condition else value_if_false
```

````python
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)  # Output: Adult
````

````python
score = 69 
status = "Passed" if score >= 65 else "Failed"
print(status)
````

````python
rows = 6
for level in range(1, rows + 1):
    print(' ' * (rows - level) + '*' * (2 * level - 1))
````

````python
count = 0
for char in "Banana":
    if char == 'a':
        count += 1
print(f"A repeats at {count} times.")
````

````python
word = "python"
rev = ""

for i in word:
    rev  = i + rev

print(rev)
````
