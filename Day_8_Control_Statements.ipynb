## **Loop Control Statements**
You can alter the flow of a loop using three keywords:

**1.** break: Terminates the loop entirely and jumps to the code immediately following the loop.

**2.** continue: Skips the remaining code inside the current iteration and jumps directly to the next iteration.

**3.** pass: Acts as a null statement or placeholder when syntax requires code, but you want nothing to execute.

```python
for i in range(1, 1001, 100):
    
    if i == 501:
        continue
    print(i)
```

```python
for num in range(5):
    if num == 4:
        break     # Stops the loop at 4
    print(num)
```

```python
for num in range(5):
    if num == 2:
        continue  # Skips printing 2
    print(num)
```

## **Loop with an else Block**
Python allows an optional else block after a loop. The else block executes only if the loop finishes naturally without encountering a break statement.

```python
for x in range(3):
    print(x)
else:
    print("Loop finished successfully!")
```

## **Accessing Indexes with enumerate()**
If you need both the item and its index position while looping over a collection, wrap the collection in enumerate().

```python
items = ["A", "B", "C"]
for index, value in enumerate(items):
    print(f"Index {index} has value {value}")
```
