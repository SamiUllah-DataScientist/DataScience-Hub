## **1. Comparison Operations**

```python
print(-18 > 90)
print(18 < 67)
print(-1000 == 1000)
print(231 != 231)
print(90 >= 78)
print(87 <= 100)
```

## **2 Boolean Operations**

```python
print(True)
print(False)
```

## **3 Logical Operations**

```python
print(True and False)
print(True or False)
print(not(False))
```
