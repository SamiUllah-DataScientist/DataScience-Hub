## **Built in Functions**
**Python** provides many built in functions to perform common tasks. Here are a few of them:

**1.** Len() 
* Returns the length of an object (string, list, etc).

```python
text = "Python"
print(len(text))
```

**2.** max()
* Returns the maximum value from a list

```python
numbers = [1,2,3,5,6]
print(max(numbers))
```

**3.** sorted()
* Sorts a list in ascending order.

```python
ratings = [4.2, 3.5, 5.0, 1.3, 3.8]
print(sorted(ratings))
```

**4.** range()
* Generates a sequence of numbers.

```python
for n in range(5):
    print(n)
```

**5.** id()
* Returns the memory address of an object.

```python
x = 10
print(id(x))
```

**6.** eval()
* Evaluates a string as a python expression.

```python
expression = "5 + 3 * 2"
result = eval(expression)
print(result)
```

## **Key Components of Function**
1. `def`:-> A keyword that makes the beginning of the function.
2. `function_name` :-> A unique name for identifying the function.
3. `parameters` :-> Values passed to the function for processing.
4. `(:) colon` :-> Marks the end of the function header.
5. `Function Body` :-> Contains the logic of the function
6. `'''docstring'''` :-> Used to describe what the function does.
7. `return` :-> Returns a result to the caller.

```python
# Defining  a function named add that takes two numbers as a parameters.
def add(a, b):
    """This function takes two numbers, adds them and returns the sum."""
    print(f"First Number is {a}")
    print(f"Second Number is {b}")
    addition = a + b
    return addition

# calling the function and storing the result
res = add(8, 7)
print(f"Result = {res}")
print(add.__doc__)
```

### **2. Parameters vs. Arguments**

Parameters are the placeholders defined in the function header.

Arguments are the actual values passed to the function when it is called

```python
def greet(name= "Mahnoor"):  # parameters
    print(f"Hello {name}")

greet("Sami") # arguments
greet("Ali")
greet()
greet("Qasim")
```

```python
def greet(name):  # parameters
    print(f"Hello {name}")

greet("Sami") # arguments
greet("Ali")
greet("Qasim")
```

```python
def student_info(name, age, field):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Field: {field}")

student_info("Sami", 21, "Data Science")
```

### **3. Returning Values**
Functions can send data back to the caller using the return statement. Once a return statement is reached, the function exits immediately. If no return statement is specified, the function returns None by default

```python
def add_numbers(a, b):
    return a + b  # Returns the sum to the caller

result = add_numbers(5, 7)
print(result)  # Output: 12
```

```python
def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average

marks = [90, 83, 76, 53]
avg = calculate_average(marks)
print(avg)
```

```python
def calculate_total(numbers):
    total = 0

    for number in numbers:
        total += number

    return total

sales = [90, 83, 76, 53]
total_sales = calculate_total(sales)
print(total_sales)
```

```python
def performance_status(average):
    if average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Average"
    else:
        return "Needs Improvement"

status = performance_status(84)

print(status)
```

### **Keyword Arguments**

```python
def student(name, age, city):
    print(name, age, city)

student(
    city="Layyah",
    name="Sami",
    age=21
)

student("Sami", 21, "Layyah")
```

```python
def analyze_marks(marks):
    total = sum(marks)
    average = total / len(marks)
    highest = max(marks)
    lowest = min(marks)

    return total, average, highest, lowest

total, average, highest, lowest = analyze_marks(marks)

print(total, average, highest, lowest)
```

### **🎯 TASK 4 — Performance Function**

```python
def get_performance_status(average):
    if average >= 90:
        return "Outstanding"
    elif average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Average"
    else:
        return "Needs Improvement"

status = get_performance_status(67)
print(status)
```
