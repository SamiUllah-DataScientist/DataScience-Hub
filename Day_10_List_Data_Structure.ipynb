## **1. List** []
In Python, a **list** is a built-in, mutable, and ordered collection of items enclosed in square brackets []. Lists can hold duplicate values and store multiple data types simultaneously, such as integers, strings, and booleans.

```python
empty_list = []
print(empty_list)

# CHECKS IF IT IS A LIST STRUCTURE OR NOT
print(type(empty_list))
```

```python
my_list = ['Sami', 23, 'Data Science', 76000, 'VU', True, 58.93]

print("------- Introduction List --------")
print(type(my_list))

print(my_list)
```

## **2. List Constructor**

```python
new_list = list("abc")
print(new_list)

print(" ".join(new_list))
```

## **3. Indexing & Slicing**

```python
my_list = ['Sami', 23, 'Data Science', 76000, 'VU', True, 58.93]

# indexing -> positive

print("Age:", my_list[1])
print("University Name:", my_list[4])

# indexing -> negative

print("Salary:", my_list[-4])
print("Is Student:", my_list[-2])
```

```python
# slicing -> positive
my_list = ['Sami', 23, 'Data Science', 76000, 'VU', True, 58.93]
print(my_list[:5])
print(my_list[3:])

# slicing -> negative
print(my_list[-4:-2])
# invert all list elements
print(my_list[::-1])
```

## **4. List Methods**

```python
fruits = ['apple', 'mango', 'grapes']
# 1. checks length of list
print(len(fruits))

# 2. add another element at last
fruits.append('orange')
print(fruits)
print(type(fruits))

# 3. add at specific index
fruits.insert(2,'matermelon')
print(fruits)

# pop the element from list
fruits.pop(3)
print(fruits)

# 5. remove another method
fruits.remove('mango')
print(fruits)
print(len(fruits))
```

## **5. Sorting & Reversing**

```python
my_list = ['Sami', 23, 'Data Science', 76000, 'VU', True, 58.93]

# reverse through thins method
print(my_list[::-1])

new_list = [100, -9, 7, 10, 367, 223, 32, 0 , 18, -56]
print(new_list)
odered_list = sorted(new_list)
print(odered_list)

# reverse order like descending order
new_list.sort(reverse=True)
print(new_list)
```

```python
cart = ["T-shirt", "Lamp", "Pen"]
fav_items = ["Pen","Headphones", "Phone"]

# Add all the items from fav_items to cart
cart.extend(fav_items)

print(cart) 
```

```python
marks = [13, -9, 78, 1000, 67, -32, 87, 5, 3232]

print("Maximum Element:", max(marks))
print("Minimum Element:", min(marks))
```

```python
print(sum(marks))
print("Average:", sum(marks) / len(marks))
```
