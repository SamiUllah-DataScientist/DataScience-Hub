## **Data Structures in Python**
Python provides **four** built-in data structures—lists, dictionaries, sets, and tuples—alongside powerful advanced structures in its standard library and the ability to build custom user-defined structures.

```python
## Python Collections as a Data Structure
# USed to store multiple values in one variable

# List: Ordered and changeable. Allows duplicate items.
# Tuple: Ordered and unchangeable. Allows duplicate items.
# Set: Unordered and unindexed. Contains unique values only (no duplicates).
# Dictionary: Stores data in key-value pairs.
```

## **1. List**
**List** is an Ordered, mutable collections that allow duplicate elements.


```python
fruits = ["Apple", "Grapes", "Banana", "Orange"]

print(fruits)
print(type(fruits))
```

## **2. Tuple**
**Tuple** is an Ordered, immutable collections that allow duplicate elements.

```python
colors = ("Red","Green","Blue")

print(colors)
print(type(colors))
```

## **3 Set**
**Set** is an Unordered collections of unique elements with no duplicates allowed.

```python
numbers = {1, 3, 5, 6, 6, 5, 3, 8, 9, 0}

# print only repeat numbers first time
print(numbers)
print(type(numbers))
```

## **4. Dictionary**
**Dictionary** is an Unordered (but preserves insertion order in Python 3.7+), mutable mappings of unique keys to values.

```python
intro_dict = {
    'name' : "Sami",
    'skill' : "Python",
    'age' : 18
}

print(intro_dict)
print(type(intro_dict))
```

```python
# Using curly braces
user = {
    "name": "Alice",
    "age": 28,
    "city": "New York"
}

print(user)
# Using the dict() constructor
user = dict(name="Alice", age=28, city="New York")

# Empty dictionary
empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```
