## **Local Variable**
A local variable are those defined within a function. They are only accessible within the function and cannot be assessed outside of it to help visualise this consider the example of class white board.

```python
# Example 
def my_fun():
    y = 10 # local variable
    print("Inside function: x =", x, "y =", y)

x = 20 # global variable
my_fun()
```

## **Global Variable**
Global variable are defined outside of functions and are accessible throughout the entire programme in our analogy the schools notice board can be seen by everyone in the school not just those in a specific class similarly a global variable can be accessed by any function in the program.

```python
# Example
x = 10
def my_fun():
    y = 5
    print("Inside function: x =", x, "y =", y)

my_fun()
print("Outside function: x =", x)
```

## **Real Time Concept**

```python
x = 10 # global variable
def my_fun():
    y = 5 # local variable
    print("Inside function: x =", x, "y =", y)

my_fun()
print("Outside function: x =", x)

# Trying to assess a local variable Y outside the function will result in an error.
# print("Outside function: y =", y) # This will raise an error because y is not defined outside the function.
```

### **LEGB Rule in Action**

```python
x = "global"
def outer_function():
    x = "enclosing"
    def inner_function():
        x = "local"
        print("Inside inner_function: x =", x)
    inner_function()
    print("Inside outer_function: x =", x)

outer_function()
print("Outside function: x =", x)
```

### **Modify Global Variable**
To modify a global variable inside a function, we must use the global keyword.

```python
counter = 0 # global variable
def increment():
    global counter
    counter += 1

increment()
print("Updated Counter:", counter)
```
