## **Types of Variable**
There are different types of variables are available in python as same as in other languages like string, integer, float and boolean.

## **1. Integer Data Type**
In Python, the **int** data type represents integers, which are whole numbers without any decimal places. They can be positive, negative, or zero.

```python
x = 10          # Positive integer
y = -5          # Negative integer
z = 0           # Zero

print(type(x))  # checks its type
```

```python
# int Constructor
price = int(450.98)

print(price)
print(type(price))

print(int("42"))      # Output: 42 (converts string to integer)
print(int("1010", 2)) # Output: 10 (converts binary string to base 10 int)
```

## **2. Float Data Type**
In Python, the **float** data type represents real numbers containing one or more decimals.

```python
weight = 76.90

print(weight)
print(type(weight))
```

```python
# Literal assignment
price = 19.99 
exact_zero = 5.0

print(price)
print(exact_zero)
# Scientific notation 
large_num = 2.5e2 
small_num = 3.14e-3

print(large_num)
print(small_num)
# Constructor conversion
string_to_float = float("3.14")
int_to_float = float(42) 

print(string_to_float)
print(int_to_float)
```

```python
print(5 + 2.0)  
print(10 / 2)  
```

```python
val = 4.5
print(val.is_integer())        # False
print(val.as_integer_ratio())  # (9, 2)
print(val.hex())               # '0x1.2000000000000p+2'
```

```python
# Expected: 0.3
print(0.1 + 0.2)  # wow that's magic
```

## **3. String Data Type**
A **string** is an immutable, ordered sequence of Unicode characters used to store and manipulate textual data.

```python
name = "Sami"

print(name)
print(type(name))
```

```python
# Multiline string
multiline = """This string spans

multiple lines."""

print(multiline)

print(type(multiline))
```

## **4. Boolean Data Type**
In Python, the **Boolean** data type (bool) represents one of two built-in truth values: True or False. 

```python
is_student = True

print(is_student)
print(type(is_student))
```

```python
print(bool(""))          # Output: False (Empty string)
print(bool(0))           # Output: False (Zero integer)
print(bool([]))          # Output: False (Empty list)

print(bool("Python"))    # Output: True (Non-empty string)
print(bool(42))          # Output: True (Non-zero number)
```
