                   print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)

```python
# print Welcome Message
print("Hello World")
```

## **String Indexing**
**String indexing** is the process of accessing specific individual characters within a string based on their position. 

```python
message = "Wishing you a fast and smooth recovery."
print("Message:",message)
# positive indexing
print(message[4])
print(len(message))
print(message[19])

# negative indexing
print("Last Alphabet:",message[-1])
print(message[-38])
```

## **String Slicing**
**String slicing** in Python is a technique used to extract a specific portion (substring) of a string by utilizing index positions.

The core syntax for string slicing is            
                                        
                                        string[start:end:step]

```python
message = "Wishing you a_fast and smooth recovery."
# string slicing
print(message)
print(message[3:9])
print(message[13:28])
print(message[:13])
print(message[21:])
# negative slicing
print(message[8:-2])
```

## **String Methods**
**String methods** are built-in functions provided by programming languages to manipulate, format, and inspect text.
### **1. Change Case Methods**

```python
text = "hello WOrlD"
# Changing Case in all lower case
print("Lower Case:",text.lower())
# changing Case in all upper case
print("Upper Case:",text.upper())
# changing case in proper text format
print("Title Case:",text.title())
# capitalize case text
print("Capitalize Case:",text.capitalize())
# inverts the string case 
print("Swapcase Case:",text.swapcase())
# Aggressive lowercase conversion
print("Casefold Case:",text.casefold())
```

### **2.Splitting & Joining**

```python
phrase = "         clean me    "
print("Uncleaned Text:",phrase)
# first of all lenth of text
print("Length:",len(phrase))
clean_phrase = phrase.strip()
print("Clean Text:",clean_phrase) 

# spliting text
line = "Be yourself; everyone else is already taken."
print("Text in List:",line.split())
list_text = line.split()

join_text = ["My","name","is","Abdullah"]
print("-".join(join_text))

# replace text
my_text = "My Name is Abdullah"
print("Real Line:", my_text)
# find name
print("Find Name:",my_text.find('Abdullah')) #if not found it returns -1
replace_text = my_text.replace('Abdullah','Sami')
print("Original Line:",replace_text)
```

### **3. Searching & Counting**

```python
fruit = "Banana"
# counts a in the text
print("\'A\' Repeats a Time:",fruit.count('a'))
# find the index of th character
print("\'n\' is at index:",fruit.index('n'))
# checks a string start with A
print(fruit.startswith('A'))
# checks a string endswith A
print(fruit.endswith('a'))
```

## **4. Inspection Properties**

```python
print("Python".isalpha())    # True
print("Python 3".isalpha())  # False (contains a space and a digit)
print("".join([]).isalpha()) # False (empty string)

print("_"* 15)
print("12345".isdigit()) # True
print("12.34".isdigit()) # False (contains a decimal point)
print("-45".isdigit())   # False (contains a negative sign)

print("_"* 15)
print("Agent007".isalnum()) # True
print("Agent 007".isalnum()) # False (contains a space)
print("Cool!".isalnum())     # False (contains an exclamation mark)

print("_"*15)
print("   ".isspace())   # True
print("\n\t".isspace())  # True
print(" code ".isspace()) # False (contains letters)
```

### **4. Formatting & Padding**

```python
# f-sring -> most famous format of string in python
name = "Sami"
f_name = "Muhammad Ramzan"
print(f"My name is {name}. \nMy Father Name is {f_name}.")

print("_"*25)
# without f string
print("My Name is", name)
print("My Father Name is", f_name)


# most fancy Style of text formatting is using .format function.

print("My Name is {} and Father Name is {}".format('Sami','Muhammad Ramzan'))
```

```python
# Padding an integer/string to width 5
number_str = "42"
print(number_str.zfill(5))

# Handling negative signs
negative_str = "-7"
print(negative_str.zfill(4))

# Centering with default space padding
title = "Menu"
print(title.center(10))

# Centering with a custom fill character
print(title.center(10, "="))
```
