### **📊 Task — Student Academic Analyzer**

percentage >= 80 → Excellent
percentage >= 70 → Good
percentage >= 60 → Average
else → Needs Improvement

```python
std_name = input("Enter the Student Name: ")

matric_obtained_marks = int(input("Enter the Obtained Matric Marks: "))
matric_total_marks = int(input("Enter the Total Matric Marks: "))

inter_obtained_marks = int(input("Enter the Obtained Intermediate Marks: "))
inter_total_marks = int(input("Enter the Total Intermediate Marks: "))

attendance_percentage = int(input("Enter your attendance percentage (0-100): "))

matric_percentage = (matric_obtained_marks / matric_total_marks) * 100
inter_percentage = (inter_obtained_marks / inter_total_marks) * 100

average_percentage = (matric_percentage + inter_percentage) / 2

if attendance_percentage >= 75:
    attendance_status = "Good"
else:
    attendance_status = "Warning"

if average_percentage >= 80:
    academic_status = "Excellent"
elif average_percentage >= 70:
    academic_status = "Good"
elif average_percentage >= 60:
    academic_status = "Average"
else:
    academic_status = "Improvement Needed"

print("=" * 40)
print("       STUDENT ACADEMIC REPORT")
print("=" * 40)

print(f"Student: {std_name}")

print(f"Matric Percentage: {matric_percentage:.2f}%")
print(f"Intermediate Percentage: {inter_percentage:.2f}%")
print(f"Average Percentage: {average_percentage:.2f}%")

print(f"Attendance: {attendance_percentage}%")
print(f"Attendance Status: {attendance_status}")

print(f"Academic Status: {academic_status}")

print("=" * 40)
```

```python
salaries = [45000, 62000, 38000, 75000, 55000, 48000, 90000, 42000, 68000, 51000]

total = 0
highest_salary = salaries[0]
lowest_salary = salaries[0]
earning_above_60000 = 0
earning_less_50000 = 0

for salary in salaries:
    total += salary
    if salary > highest_salary:
        highest_salary = salary
    if salary < lowest_salary:
        lowest_salary = salary
    if salary >= 60000:
        earning_above_60000 += 1
    if salary < 50000:
        earning_less_50000 += 1

average_salary = total/len(salaries)
count = 0
for salary in salaries:
    if salary > average_salary:
        count += 1

print(f"Total Salary: {total}")
print(f"Average Salary: {average_salary:.2f}")
print(f"Highest Salary: {highest_salary}")
print(f"Lowest Salary: {lowest_salary}")
print(f"Employees Earning >= 60000: {earning_above_60000}")
print(f"Employees Earning < 50000: {earning_less_50000}")
print(f"Salaries Above Average: {count}")
```

```python
import math
print("----- Area Calculation -----")
print("Press 1 to calculate Area of Square")
print("Press 2 to calculate Area of Rectangle")
print("Press 3 to calculate the Area of Circle")
print("Press 4 to calculate the Area of Triangle")

choice = int(input("Enter Your Choice: "))

if choice == 1:
    length = float(input("Enter the length of Square: "))
    area_of_sqr = length * length

    print(f"\nThe Area of Square is {round(area_of_sqr,4)}.")
elif choice == 2:
    length = float(input("Enter the length of Reactangle: "))
    width = float(input("Enter the Width of Rectangle: "))

    area_of_rec = length * width

    print(f"\nThe Area of Rectangle is {round(area_of_rec,4)}.")

elif choice == 3:
    rad = float(input("ENter the radius of Circle: "))

    area_of_circle = math.pi * rad ** 2

    print(f"\nThe Area of Circle is {round(area_of_circle,5)}.")

elif choice == 4:
    base = float(input("Enter the base of Triangle: "))
    height = float(input("Enter the height of Triangle: "))

    area_of_triangle = (base * height) / 2

    print(f"\nThe Area of Triangle is {round(area_of_triangle,4)}.")
else:
    print("Thanks for Using my Area Calculator...")
```
