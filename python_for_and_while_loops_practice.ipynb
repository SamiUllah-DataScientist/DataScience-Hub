## **Loops in Python**
Python features two primary types of loops: for loops (used to iterate over a sequence or fixed range) and while loops (used to repeat code as long as a condition remains true).

## **for Loop**
Use a for loop when you know in advance how many times you want to execute a block of code, or when iterating over a collection (like a list, tuple, dictionary, or string).

```python
for i in range(3):
    print(i)
```

```python
n = 1
for n in range(11):
    print("Hello World.")
```

```python
for i in range(11, 21):
    print(i, end=' ')
```

```python
for k in range(6, 51, 2):
    print(k, end = ' ')
```

```python
tab_no = int(input("Enter a Number: "))
for i in range(1, 11):
    print(f"{tab_no} x {i} = {tab_no * i}")
```

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit.capitalize())
```

```python
items = ["A", "B", "C"]
for index, value in enumerate(items):
    print(f"Index {index} has value {value}")
```

## **while Loop**
Use a while loop when you need to repeat an action until a specific condition becomes false. You must manually modify the condition variable inside the loop to avoid an infinite loop.

```python
count = 0
while count < 3:
    print(count)
    count += 1  # Increments count so the condition eventually becomes False
```

```python
n = int(input("Enter the Number: "))
stop_no = 1
while stop_no <= 10:
    print(f"{n} x {stop_no} = {n * stop_no}")
    stop_no += 1
```

## **Nested Loops**
You can place loops inside other loops. The inner loop runs completely for every single iteration of the outer loop.

```python
for outer in ["First", "Second"]:
    for inner in [1, 2]:
        print(outer, inner)
```

```python
for o in range(1,3):
    print(o, "Outer Loop:")
    for i in range(1, 4):
        print("       ", i, "Inner Loop")
```

## **Pattern Solving**

```python
for n in range(1,6):
    for i in range(n):
        print(n, end = ' ')
    print()
```

```python
for n in range(1,6):
    for i in range(n):
        print("*", end = ' ')
    print()
```

```python
for n in range(1,6):
    for i in range(n):
        print(i+1, end= ' ')
    print()
```

```python
sales = [1200, 4500, 7800, 3200, 9000]

list_above_5000 = []
count_above_5000 = 0
for sale in sales:
    print(f"Sale: {sale} PKR")
    if sale > 5000:
        list_above_5000.append(sale)
        count_above_5000 += 1
print("\nSales Above 5000:")
for sale in list_above_5000:    
    print(sale)
print(f"\nCount: {count_above_5000}")
```

```python
sales = [1200, 4500, 7800, 3200, 9000, 6500, 4100]
days = list(range(1, 8))

for s , d in zip(sales, days):
    print(f"Day {d}: {s} PKR")
print("\n\n")
for s , d in zip(sales, days):
    if s > 5000:
        print(f"Day {d}: {s} PKR")
```

```python
sales = [1200, 4500, 7800, 3200, 9000, 6500, 4100]

for i in range(len(sales)):
    print(f"Day {i + 1}: {sales[i]} PKR")

for i in range(len(sales)):
    if sales[i] >= 5000:
        print(f"Day {i + 1}: {sales[i]} PKR")
```

```python
sales = [1200, 4500, 7800, 3200]

total = 0

for sale in sales:
    total += sale

print(total)
```

```python
sales = [1200, 4500, 7800, 3200]

total = 0

for sale in sales:
    total += sale

average = total / len(sales)

print(f"Total: {total}")
print(f"Average: {average:.2f}")
```

```python
sales = [1200, 4500, 7800, 3200, 9000]

highest = sales[0]

for sale in sales:
    if sale > highest:
        highest = sale

print(highest)
```

```python
lowest = sales[0]

for sale in sales:
    if sale < lowest:
        lowest = sale

print(lowest)
```
