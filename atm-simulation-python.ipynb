### **🏧 Project — ATM Transaction Simulator**

========== ATM ==========
1. Check Balance
2. Deposit
3. Withdraw
4. Exit

```python
balance = 50000
menu = """========== ATM ==========
1. Check Balance
2. Deposit
3. Withdraw
4. Exit"""

while True:
    print(menu)

    choice = int(input("Choose your Option: "))

    if choice == 1:
        print(f"\nCurrent Balance: {balance}")

    elif choice == 2:
        deposit_amount = int(input("Enter deposit amount: "))
        balance += deposit_amount
        print(f"You have successfully deposited {deposit_amount} moeny in account.")
    elif choice == 3:
        withdraw_amount = int(input("Enter withdraw amount: "))
        if balance >= withdraw_amount:
            balance -= withdraw_amount
            print(f"You have successfully withdraw {withdraw_amount} money.")
        else:
            print("Insufficient Balnace")
    elif choice == 4:
        print("Thanks You...")
        break
    else:
        print("Invalid Choice Entered...")
```
