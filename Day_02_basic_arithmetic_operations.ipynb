## **1. Integer Data Type**
An **integer** is a built-in data type (represented by the int class) that stores whole numbers—positive, negative, or zero—without any decimal points. 

```python
num1 = 8993
num2 = -9783
num3 = 0

# check type of variable 
print(type(num1))
print(type(num2))
print(type(num3))
```

## **2. Basic Arithmetic Operations**
Python supports all standard mathematical operations on integers:
1. Addition
2. Subtraction
3. Multiplication
4. Division
5. Floor Division
6. Modulus (Remainder)
7. Exponentiation (Power)

```python
name = input("Enter your name: ")
print("Hello", name)
```

```python
n1 = int(input("Enter 1st Number: "))
n2 = int(input("Enter 2nd Number: "))

print("1. Addition -> (+):", n1 + n2)
print("2. Subtraction -> (-):", n1 - n2)
print("3. Division -> (/):", n1 / n2)
print("4. Multiplication -> (*):", n1 * n2)
print("5. Floor Division -> (//):", n1 // n2)
print("6. Exponentiation -> (**):", n1 ** n2)
print("7. Modulus -> (%):", n1 % n2)
```

## **4. Float Data Type**
A **float** (floating-point number) is the built-in data type used to represent any number with a decimal point. 

```python
# Direct assignments
pi = 3.14159
negative_val = -10.5
whole_as_float = 5.0

print(type(pi))
# Scientific notation (equivalent to 3.0 * 10^4)
scientific = 3e4  # Results in 30000.0

print(scientific)
# Standard division always returns a float
result = 10 / 2  # Results in 5.0
print(result)
```

```python
# Convert an integer
print(float(42))      # Output: 42.0

# Convert a string
print(float("7.77"))  # Output: 7.77

# Convert a boolean
print(float(True))    # Output: 1.0
```
