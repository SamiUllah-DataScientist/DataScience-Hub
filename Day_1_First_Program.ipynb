## **Day_01 Print Function**
The **print()** function in Python is a built-in function used to send text, numbers, or other data objects directly to the console or screen.
## **Basic Syntax**
                   print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)

```python
# print Welcome Message
print("Hello World")
```

```python
greeting = "May Allah reward you with goodness"
print(greeting)
```

```python
print("Hello")
print("World")
```

```python
print("Hello", end = ' ')
print("World")
```

```python
print("Hello", "World", sep = '@')
```

```python
print("A","B","C")

print("A","B","C", sep = '-')

print("A","B","C", sep= '/')

print("A","B","C", sep= '\n')
```

```python
university = "University of Lahore"
print("University:",university)
print("-"*30)
```
