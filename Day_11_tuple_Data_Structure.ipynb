### **Tuple Data Structure**
A **tuple** is a built-in data type used to store a collection of items in a single variable. It is defined by three core characteristics: it is ordered, it allows duplicate values, and most importantly, it is immutable (cannot be changed after creation).

```python
student = ("Sami", 21, 3.69)

print(f"Tuple ELements: {student}")
print(type(student))
```

```python
# Standard tuple
coordinates = (40.7128, -74.0060)

print(f"Coordinates : {coordinates}")
# Tuple with mixed data types
user_profile = ("Alice", 30, True)

print(f"User Profile: {user_profile}")
# CRITICAL: A single-element tuple requires a trailing comma
single_item = ("apple",)  # Without the comma, Python sees this as a plain string
print(single_item)
print(type(single_item))
```

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
student = ("Sami", 21, "Layyah", 3.69)

print(student[0])
print(student[1])
print(student[2])
print(student[3])
```

```python
fruits = ("apple", "banana", "cherry", "orange")

print(fruits[0])    # Output: apple
print(fruits[-1])   # Output: orange (negative indexing starts from the back)
print(fruits[1:3])  # Output: ('banana', 'cherry') (slicing)
```

```python
student = ("Sami", 21, "Layyah", 3.69, "Data Science")

print(student[0:3])
```

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[0:6:2])
```

```python
print(numbers[::-1])
```

```python
student = ("Sami", 21, 3.69)

student[1] = 22
```

## **Tuple Unpacking**
You can easily extract the values of a tuple back into individual variables.

```python
point = (10, 20, 30)
x, y, z = point

print(x)  # Output: 10
print(y)  # Output: 20
```

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
student = (
    ("Sami", 21),
    ("Ali", 22),
    ("Hamza", 20)
)

print(student[0])
print(student[1][0])
```

## **TUPLE METHODS**

```python
numbers = (10, 20, 10, 30, 10, 40)

print(numbers.count(10))
```

```python
numbers = (10, 20, 30, 40)

print(numbers.index(30))
```

```python
numbers.index(100)
```

## **Built in Functions**

```python
numbers = (10, 20, 30, 40, 50)

print(max(numbers))
print(min(numbers))
print(sum(numbers))
print(len(numbers))
print(tuple(sorted(numbers)))
```
