## **Dictionary in Python**
A Python dictionary is a built-in data structure used to store data in key-value pairs. It is optimized for retrieving data quickly when you know the specific name (the key) rather than a numeric index. 

In Python 3.7 and later, dictionaries are ordered (they maintain the order of insertion), changeable (mutable), and do not allow duplicate keys.

### **1. Creating a Dictionary**
You can create a dictionary using curly brackets {} or the built-in dict() constructor.

```python
# Using curly brackets
user = {
    "name": "Alice",
    "age": 28,
    "is_member": True
}

print(user)
# Using the dict() constructor
user_2 = dict(name="Bob", age=32, is_member=False)

print(user_2)
```

```python
student = {
    "name": "Sami Ullah",
    "age": 21,
    "field": "Data Science",
    "cgpa": 3.69
}

print(student)

print("=" * 40)
print("       STUDENT INFORMATION PORTAL")
print("="*40)
for var in student:
    print(f"{var.title():<10}    -> {student[var]:<20}")
```

```python
marks = {
    1: 85,
    2: 92,
    3: 78
}

print(marks)
print(f"Keys: {list(marks.keys())}")
print(f"Values: {list(marks.values())}")
```

```python
student = {
    "name": "Sami",
    "age": 21,
    "cgpa": 3.69,
    "skills": ["Python", "SQL"],
    "active": True
}

for var in student:
    print(type(student[var]))
```

### **2. Accessing Values**
You can retrieve a value by placing its key inside square brackets [] or by using the .get() method.

```python
# Standard way (Raises KeyError if the key doesn't exist)
print(user["name"])  # Output: Alice

# Safer way (Returns None or a default value if the key doesn't exist)
print(user.get("email"))        # Output: None
print(user.get("email", "N/A")) # Output: N/A
```

```python
student = {
    "name": "Sami",
    "age": 21,
    "cgpa": 3.69
}

print(student["name"])
print(student["cgpa"])
print(student.get("city", "Not Available"))
```

### **3. Adding and Modifying Items**
You can add a new key-value pair or update an existing one using the assignment operator =.

```python
# Adding a new key-value pair
user["email"] = "alice@example.com"

# Updating an existing value
user["age"] = 29 
```

### **4. Removing Items**
Python provides multiple ways to delete items based on your needs:

```python
# Removes a specific key and returns its value
age = user.pop("age") 

# Removes and returns the last inserted key-value pair as a tuple
last_item = user.popitem() 

# Deletes an item directly using the 'del' keyword
del user["is_member"]

# Empties the entire dictionary
user.clear() 
```

```python
for key, value in student.items():
    print(key, value)
```

### **Nested Dictionary**

```python
student = {
    "name": "Sami",
    "academic": {
        "matric_percentage": 96.55,
        "inter_percentage": 84.5,
        "cgpa": 3.69
    }
}

print(student["academic"]["inter_percentage"])
```

```python
student = {
    "name": "Sami",
    "skills": ["Python", "SQL", "Excel"]
}

print(student["skills"])
print(student["skills"][:2])
```

```python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Sami", "marks": 95},
    {"name": "Ayesha", "marks": 90}
]

for i in students:
    print(i["name"])
```

## **Important for Data Science Projects**

```python
students = [
    {"student_id": 101, "name": "Ali", "age": 20, "marks": 85},
    {"student_id": 102, "name": "Sami", "age": 21, "marks": 95},
    {"student_id": 103, "name": "Ayesha", "age": 20, "marks": 90}
]

print("-" * 50)
print("Student_ID     | Age   | Name             | Marks")
print("=" * 50)

for i in students:
    print(f"{i["student_id"]:<14} | {i["age"]:<5} | {i["name"]:<16} | {i["marks"]}")
```
