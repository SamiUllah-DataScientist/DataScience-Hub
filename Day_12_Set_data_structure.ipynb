## **Set Data Structure**

A **set** is a built-in data type used to store an unordered collection of unique items. Sets are highly optimized for membership testing (checking if an item exists) and removing duplicate entries from other collections

```python
cities = {"Layyah", "Lahore", "Layyah", "Islamabad"}

print(cities)
```

```python
numbers = {1, 2, 2, 3, 3, 3, 4}

print(numbers)
```

```python
# Creating a set with elements
fruits = {"apple", "banana", "cherry"}

print(fruits)
# Creating an empty set (You MUST use set(), because {} creates an empty dictionary)
empty_set = set() 
print(empty_set)
print(type(empty_set))
```

## **Set Methods**

```python
numbers = {1, 2, 3}

# Add a single element
numbers.add(4)         # {1, 2, 3, 4}

# Add multiple elements using .update()
numbers.update([5, 6]) # {1, 2, 3, 4, 5, 6}

# Remove an element (Raises KeyError if not found)
numbers.remove(3)      # {1, 2, 4, 5, 6}

# Remove an element safely (Does NOT raise an error if missing)
numbers.discard(99)    # Safe to use

print(numbers)
```

## **Set Operations**

```python
python_students = {"Ali", "Sami", "Hamza", "Ayesha"}

sql_students = {"Sami", "Ayesha", "Zain", "Bilal"}

all_students = python_students.union(sql_students)

print(f"All Students: {all_students}")

common_students = python_students.intersection(sql_students)

print(f"Common Students: {common_students}")
```

```python
python_only = python_students.difference(sql_students)

print(python_only)
```

```python
result = python_students.symmetric_difference(sql_students)
# Jo kisi ek set mein hain but dono mein common nahi
print(result)
```
