## **Virtual Environment**

A **virtual environment** is an isolated workspace in python that allows you to manage dependencies for different projects separately each virtual environment has its own Python interpreter and instal packages ensuring that project do not interface with each other.

**Note:** Virtual environments allow you to instal different versions of libraries for each project without conflicts.

### **Creating Virtual Environment**

1. **Open Anacoda Prompt in OS**
   * Press the Windows Key and search for Anaconda Prompt.
   * Open the Command line Interface (CLI).
2. **Navigate the Desired Directory**
   * Use the cd (change directory) command to navigate to your project folder.
   * cd Documents or cd Python Programming
3. **Tips**
   * Use the **Tab** key to automate complte directory names and avoid spelling mistakes.
4. **Create the Enviroment**
   * Use the **conda create** command to create a virtual environment.

## **Syntax**

```python
conda create -name or -n <environment_name> python=<version>
```

**Example**

conda create -n first_env python= 3.12

5. **Confirm Installation**
   * The system will prompt you to confirm the installtion of the environment and its packages.
   * Type y (yes) to proceed.

6. **Activate the Environment**
   * Type this command "conda activate first_env"
7. **Install Packages in the Env**
   * pip install numpy=2.3 or conda install pandas
   * conda list to check the packages
8. **Deactivate the Env**
   * conda deactivate
9. **Listing All Env**
   * conda info --envs
10. **Remove Enviroment and Packages**
   * conda remove --n first_env --all
